## Staleness test at fixed reproducible wall clock

```python
import datetime as dt,importlib.util
spec=importlib.util.spec_from_file_location("monitor",r/"scripts/paper_monitor.py")
m=importlib.util.module_from_spec(spec);spec.loader.exec_module(m)
status=m.check(b["as_of"],dt.datetime(2026,11,25,5,0,tzinfo=dt.timezone.utc))
assert status["status"]=="BLOCKED_STALE_DATA" and not status["forecast_emitted"]
print(status)
```
